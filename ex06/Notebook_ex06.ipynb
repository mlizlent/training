```python
import numpy as np
from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
```

```python
#Load the California Housing dataset (no manual train/test split needed)
housing = fetch_california_housing()
X, y = housing['data'], housing['target']
```

```python
# Define the custom single train/test split for cv
custom_cv = [(np.arange(18576), np.arange(18576, 20640))]
```

```python
# Define the estimator and hyperparameter grid (at least 3 values each)
rf = RandomForestRegressor(random_state=43)
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [10, 20, None],
    'min_samples_leaf': [1, 2, 5]
}
```

```python
#  Initialize and fit GridSearchCV
gs = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=custom_cv,
    n_jobs=-1,
    verbose=1
)
gs.fit(X, y)
```

```python
# Function to extract and print the best model, parameters, and score
def select_model_verbose(gs):
    trained_model = gs.best_estimator_
    best_params = gs.best_params_
    best_score = gs.best_score_
    
    print(f"Best Score (R2): {best_score:.4f}")
    print(f"Best Hyperparameters: {best_params}")
    
    return trained_model, best_params, best_score

trained_model, best_params, best_score = select_model_verbose(gs)

```

```python
#  Predict on the new data point
new_point = np.array([[3.2031, 52., 5.47761194, 1.07960199, 910., 2.26368159, 37.85, -122.26]])
prediction = trained_model.predict(new_point)
print(f"Predicted value for new_point: {prediction[0]:.4f}")
```
